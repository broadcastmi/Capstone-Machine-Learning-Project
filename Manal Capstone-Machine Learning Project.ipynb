```python
##Examine your dataset
#Import Pandas and load your dataset
import pandas as pd
import numpy as np

df = pd.read_excel("Public_Transport_Trips_EDA.xlsx")

#Examine the first five rows
df.head()

#Examine the last five rows
df.tail()

#Examine dataset dimensions
df.shape

#Examine column names
df.columns

#Examine data types and non-null values
df.info()

#Examine missing values
df.isnull().sum()

#Examine unique values
df.nunique()
df['Mode_of_Transport'].unique()
df['Departure_Station'].unique()
df['Arrival_Station'].unique()

#Examine frequency distributions
df['Mode_of_Transport'].value_counts()
df['Day_of_Week'].value_counts()
df['Departure_Station'].value_counts()

#Examine numerical descriptive statistics
df.describe()
df.describe(include='all')

#Examine data types individually
df.dtypes

#Examine duplicate records
df.duplicated().sum()
df['Trip_ID'].duplicated().sum()

#Examine a random sample
df.sample(5, random_state=42)

pd.set_option('display.max_columns', None)

pd.set_option('display.max_rows', 20)

df.head(20)
```

```python
#Remove empty columns and investigate unnamed columns
# Remove completely empty columns
df = df.dropna(axis=1, how='all')

# Identify columns without proper names
unnamed_columns = [
    col for col in df.columns
    if str(col).startswith("Unnamed")
]

# Examine their contents
for col in unnamed_columns:
    print(df[col].value_counts(dropna=False))

# Remove unidentified columns only after checking
df = df.drop(columns=unnamed_columns)
```

```python
#Check for missing values
# Count missing values in each column
df.isnull().sum()

# Calculate percentage of missing values
(df.isnull().sum() / len(df)) * 100

# Display rows containing missing values
df[df.isnull().any(axis=1)]
```

```python
#Check and remove duplicates
# Count duplicate rows
df.duplicated().sum()

# Display duplicate rows
df[df.duplicated()]

# Check duplicate trip IDs
df['Trip_ID'].duplicated().sum()

# Remove duplicate rows if necessary
df.drop_duplicates(inplace=True)
```

```python
#Convert data types
# Convert numerical columns
# Convert numerical columns
numeric_columns = [
    'Passenger_Count',
    'Fare_Amount',
    'Trip_Duration_Minutes'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Convert Trip_Date from Excel date format
if pd.api.types.is_numeric_dtype(df['Trip_Date']):
    df['Trip_Date'] = pd.to_datetime(
        df['Trip_Date'],
        unit='D',
        origin='1899-12-30',
        errors='coerce'
    )
else:
    df['Trip_Date'] = pd.to_datetime(
        df['Trip_Date'], errors='coerce'
    )
```

```python
#Clean departure time
if pd.api.types.is_numeric_dtype(df['Departure_Time']):
    departure_dt = pd.to_datetime(
        df['Departure_Time'] % 1,
        unit='D',
        errors='coerce'
    )
else:
    departure_dt = pd.to_datetime(
        df['Departure_Time'],
        errors='coerce'
    )

# Extract departure hour
df['Departure_Hour'] = departure_dt.dt.hour

# Extract formatted departure time
df['Departure_Time_Clean'] = departure_dt.dt.strftime('%H:%M:%S')

# Examine the results
df[['Departure_Time_Clean', 'Departure_Hour']].head()
```

```python
#Identify invalid values
# Check for negative passenger counts
df[df['Passenger_Count'] < 0]

# Check for negative fares
df[df['Fare_Amount'] < 0]

# Check for negative trip durations
df[df['Trip_Duration_Minutes'] < 0]
```

```python
#UNIVARIATE ANALYSIS
#Passenger count distribution
sns.histplot(data=df, x='Passenger_Count', bins=20, kde=True)

plt.title('Distribution of Passenger Counts')
plt.xlabel('Number of Passengers')
plt.ylabel('Frequency')

plt.show()

#Fare distribution
sns.histplot(data=df, x='Fare_Amount', bins=20, kde=True)

plt.title('Distribution of Transportation Fares')
plt.xlabel('Fare Amount')
plt.ylabel('Frequency')

plt.show()

#Trip duration distribution
sns.histplot(data=df, x='Trip_Duration_Minutes', bins=20, kde=True)

plt.title('Distribution of Trip Durations')
plt.xlabel('Trip Duration (Minutes)')
plt.ylabel('Frequency')

plt.show()

#For fare
sns.boxplot(y=df['Fare_Amount'])

plt.title('Fare Amount Boxplot')
plt.show()

#For trip duration
sns.boxplot(y=df['Trip_Duration_Minutes'])

plt.title('Trip Duration Boxplot')
plt.show()


##BIVARIATE ANALYSIS
#Passenger count versus fare amount
sns.scatterplot(
    data=df,
    x='Passenger_Count',
    y='Fare_Amount'
)

plt.title('Passenger Count vs Fare Amount')
plt.xlabel('Passenger Count')
plt.ylabel('Fare Amount')

plt.show()

#Trip duration versus fare amount
sns.scatterplot(
    data=df,
    x='Trip_Duration_Minutes',
    y='Fare_Amount'
)

plt.title('Trip Duration vs Fare Amount')
plt.xlabel('Trip Duration')
plt.ylabel('Fare Amount')

plt.show()

#sns.boxplot(
data=df,
 x='Mode_of_Transport',
 y='Fare_Amount'
)

plt.title('Fare Amount by Transportation Mode')
plt.show()

#Transportation mode versus passenger count
sns.boxplot(
    data=df,
 x='Mode_of_Transport',
 y='Passenger_Count'
)

plt.title('Passenger Count by Transportation Mode')

plt.show()

#Day of week versus number of trips
sns.countplot(
 data=df,
 x='Day_of_Week'
)

plt.title('Number of Trips by Day of Week')

plt.xticks(rotation=45)

plt.show()


##MULTIVARIABLE ANALYSIS
#Correlation matrix
correlation_matrix = df[
    ['Passenger_Count',
     'Fare_Amount',
     'Trip_Duration_Minutes']
].corr()

print(correlation_matrix)

#Correlation heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Transportation Variables Correlation Heatmap')

plt.show()

#Analyze transportation mode, day of week, and fare
df.groupby(
    ['Mode_of_Transport', 'Day_of_Week']
)['Fare_Amount'].mean()

```
