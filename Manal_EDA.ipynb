# Data Visualization

```python
sns.countplot(df['Gender'])
plt.title('Gender Count')
plt.xlabel('Number of Students')
plt.ylabel('Gender')
plt.show()
plt.savefig('genderchart.jpeg')
```

```python
gender_count = df['Gender'].value_counts()
```

```python
sns.barplot(y=gender_count.index, x=gender_count.values)
plt.title('Gender Count')
plt.xlabel('Number of Students')
plt.ylabel('Gender')
plt.show()
plt.savefig('genderchart2.jpeg')
```

```python
#sns.lineplot
sns.distplot(df['Age'])
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.show()
plt.savefig('agechart.jpeg')
```

```python
sns.lmplot(data=df, x='Age', y='Sem_GPA', col='Gender')
plt.title('Age vs GPA')
plt.xlabel('Age')
```

```python
sns.heatmap(df.select_dtypes(include='number').corr(), cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

# Rule of Thumbs for Visualization

* Numerical, Numerical = Scatterplot/heatmap

* 1 Numerical , 1 Categrical = Barchart

* 1 categorical = Countplot

* 1 Numerical = Histogram (distplot, kdeplot, histplot)
